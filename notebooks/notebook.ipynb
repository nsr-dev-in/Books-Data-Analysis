```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Importing Data

```python
df = pd.read_csv("Books_Data_Clean.csv")
```

VIEW : Top 5 Rows

```python
df.head()
```

CHECK : Summary of Dataset  

```python
df.describe()
```

FIX : Publishing Year min value (- 560.000000)

```python
df = df[df["Publishing Year"] > 1900]
```

```python
df.describe()
```

CHECK : Missing Values

```python
df.isna().sum()
```

Remove : Missing Value Rows

```python
df.dropna(subset="Book Name",inplace = True)
```

```python
df.isna().sum()
```

CHECK : Duplicate Records/Rows

```python
df.duplicated().sum()
```

CHECK : Unique Records 

```python
df.nunique()
```

CREATE : Histogram of Publishing Year

```python
plt.hist(df["Publishing Year"])
plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Distribution of Publishing Year")
plt.show()
```

CREATE : Bar Chart For Number of Books  

```python
df["genre"].value_counts().plot(kind="bar")
plt.xlabel("Genre")
plt.ylabel("No. of Books")
plt.title("No. of Books in Each Genre")
plt.show()
```

CREATE : Box Plot For Average Books Rating by Author adn Each Genre

```python
df.groupby("Author")["Book_average_rating"].mean().sort_values(ascending=False)
```

```python
sns.boxplot(x= "genre" ,y="Book_ratings_count",data=df)
plt.xlabel("Genre")
plt.ylabel("Book Ratings Count")
plt.title("Box Plot of Book Rating Count For Each Genre")
plt.show()
```

CREATE : Scatter Plot Sales Price vs Units Sold

```python
plt.scatter(df["sale price"],df["units sold"])
plt.xlabel("Sales Price")
plt.ylabel("Units Sold")
plt.title("Scatter Plot of Sale Price vs Units Sold")
plt.show()
```

CREATE : Bar Chart for Language Distribution of Books

```python
language_counts = df["language_code"].value_counts()
```

```python
plt.pie(language_counts,labels=language_counts.index,startangle=90,autopct="%1.1f%%")
plt.title("Language Distribution of Books")
plt.show()
```

```python
df.groupby("Publisher ")["publisher revenue"].sum().sort_values(ascending=False)
```

CHECK : Top Author Rating by Rating Counts

```python
df.groupby("Author_Rating")["Book_ratings_count"].mean().sort_values(ascending=False).head(1)
```

CHECK : No. of Books in Each 

```python
df.groupby("language_code").size().sort_values(ascending=False)
```

```python
df.groupby("Author_Rating")["Book_ratings_count"].var()
```

CREATE : Scatter Plot of Book Average Rating vs Book Rating Counts

```python
plt.scatter(df["Book_average_rating"],df["Book_ratings_count"])
plt.xlabel("Book Average Rating")
plt.ylabel("Book Rating Counts")
plt.title("Scatter Plot of Book Average Rating vs Book Rating Counts")
plt.show()
```

CREATE : Bar Graph for Total Gross Sales By Author

```python
total_gross_sales_by_author = df.groupby("Author")["gross sales"].sum()
```

```python
total_gross_sales_by_author.sort_values(ascending=False).head(20).plot(kind="bar")
plt.title("Total Gross Sales By Author")
plt.xlabel("Author")
plt.ylabel("Total Gross Sales")
plt.show()
```

```python
df.groupby("Publishing Year")["units sold"].sum().plot(kind="line")
plt.xlabel("Publishing Year")
plt.ylabel("Total Units Sold")
plt.title("Total Units Sold Ove The Years")
plt.show()
```
